# CS 8630 · Classroom Demo Runner
Run any demo and see its **built explanation** inline. For the *live* canvases and real animations (fl03!), use a terminal instead: `python run.py` from this folder.

Run the setup cell once, then any demo cell.

```python
import subprocess, sys, os, tempfile, glob
from IPython.display import Image, display
DEMOS = os.path.dirname(os.path.abspath('__file__'))
def show(path, data=None):
    'Run a demo headless and display its built explanation inline.'
    out = tempfile.mkdtemp()
    args = [sys.executable, os.path.join(DEMOS, path), '--save', out]
    if data: args += ['--data', data]
    r = subprocess.run(args, cwd=os.path.dirname(os.path.join(DEMOS, path)), capture_output=True, text=True)
    comps = glob.glob(os.path.join(out, '*_explained.png'))
    if comps: display(Image(comps[0], width=1150))
    else: print(r.stdout[-800:], r.stderr[-800:])
print('ready — show("flow/fl04_message_wind.py") etc. For LIVE interactive runs use a terminal: python run.py')
```
